### NumPy

**Ejercicio 1:**

1. Crear un vector con valores del 1 al 120.
2. Crear una función recibe_vector que reciba el vector y, dentro de la función, eleve al cuadrado aquellos números divisibles por 3, y regrese esos números en un nuevo vector.
3. Redimensionar el vector que devuelve la función anterior en una matriz de 20×2.
4. Crear una nueva matriz de 2×10 y multiplicarla por la matriz anterior.
5. Obtener la transpuesta de la matriz resultante.

```python
import numpy as np
```

```python
""" crear el arreglo de 1 a 120"""
vector = np.arange(1, 121,)
vector
```

```python
def recibe_vector(array):
    """ crear una nueva lista filtrando numeros cuyo residuo 3 de 0,
    luego elevarlos al cuadrado y finalizar aplicando la funcion"""
    nuevo_vector = array[array%3 == 0]
    return nuevo_vector ** 2
new_vector = recibe_vector(vector)
print(new_vector)
```

```python
"""darle forma con 20 filas,
dos columnas"""
reshaped = new_vector.reshape(20,2)
reshaped
```

```python
""" crear la matriz de 1 a 20,
con forma de 2 filas y 10 columnas para
poder multiplicar con reshaped"""
new_matrix = np.arange(1,21).reshape(2,10)
new_matrix
```

```python
"""aplicamos multiplicacion a las matrices"""
multiplicacion = reshaped @ new_matrix
multiplicacion
```

```python
transpuesta = multiplicacion.transpose()
print(transpuesta)
```

**Ejercicio 2:**

Se desea calcular el índice de masa corporal (IMC) de jugadores de un equipo de futbol. Los datos proporcionados corresponden a las listas de los nombres, altura y peso, tal como se muestra a continuación

```python
nombre= ['Bonifacio de Cifuentes',
'Ángel Machado Lara',
'Geraldo de Atienza',
'Onofre Posada Grande',
'Samu Cantón-Figuerola',
'Camilo Pol',
'Toni Galvez Lara',
'Lucho Manzano Rivas',
'Abraham Aroca',
'Anselmo Bartolomé Folch',
'Toni Inocencio Rozas Porras',
'Fausto Cortés-Arco',
'Baltasar Carreño Isern']

altura = [180, 215, 210, 210, 188, 176, 209, 200, 210, 188, 176, 209, 200]
peso = [69, 74, 72, 75, 68, 70, 71, 73, 69, 74, 72, 75, 68]
```

El valor de IMC por cada jugador resulta de aplicar la fórmula:

$$IMC = \frac{peso(kg)}{altura(m)^2}$$

**Nota:** Los valores de altura están dados en centímetros por lo que deberá hacer laconversión a metros previo a la aplicación de la fórmula. Adicionalmente, una vez obtenido el IMC se solicita:

> a) Presentar el icm para cada jugador.

> b) Indicar en qué índice se encuentra el icm más alto y a qué jugador le pertenece.

> c) Calcular el promedio de los icm resultantes.

> d) Muestre los nombres de los jugadores en orden de su imc, desde el que tiene menorimc al que tiene mayor imc.

```python
# creamos los arreglos de numpy para cada lista
nombres = np.array(nombre); alturas = np.array(altura); pesos = np.array(peso)
# creamos la conversion de centimetros a metros
conversion = (alturas/100)
# Redondeamos con np,round()
metros = np.round(conversion,2)
metros
```

```python
# crear la conversion y hacer un ciclo for con zip para alinear los nombres con su debido IMC
IMC = pesos / alturas
IMC
for nombre, imc in zip(nombres, IMC):
    print(f"{nombre}, {imc}")
```

```python
# Encontrar el indice con el valor mas alto, con np.argmax()
indice_alto = np.argmax(IMC)
print(f"Indice del IMC mas alto: {indice_alto}")
print(f"Pertenece al jugador: {nombres[indice_alto]}, cuyo IMC es de: {IMC[indice_alto]}")
```

```python
#Hallar el promedio con .mean()
promedio_imc = IMC.mean()
print(f"El promedio de IMC de los jugadores es de: {promedio_imc}")
```

```python
mas_altos = np.argsort(IMC) # devuelve los iondices de menor valor a mayor valor
names = nombres[mas_altos] # aplicamos la mascara de para encontrar el indice con el valor mas bajo dentro de los nombres, filtrando por IMC
for name, valor in zip(names, IMC[mas_altos]):
    print(f"{name}, {valor}")
```
