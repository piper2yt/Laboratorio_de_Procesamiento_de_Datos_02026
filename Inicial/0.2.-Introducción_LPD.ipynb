## 1. ¿Qué es el Procesamiento de Datos?

El **Procesamiento de Datos** es el conjunto de métodos, técnicas y herramientas que permiten **extraer, organizar, comprender, limpiar, transformar y analizar datos** para convertirlos en información útil y confiable.

Los datos pueden provenir de archivos de texto, CSV, XLS/XLSX, JSON, XML, archivos geoespaciales SHP, imágenes, bases de datos, APIs o sensores. Antes de utilizarlos en un análisis o modelo, es necesario conocer su estructura, evaluar su calidad y preparar una representación adecuada.

> **En términos simples:** el procesamiento de datos transforma datos crudos, heterogéneos y posiblemente incompletos en datos consistentes, interpretables y listos para el análisis.

### ¿Por qué es tan importante?

La calidad de cualquier análisis depende de la calidad de los datos que lo sustentan. Un procesamiento cuidadoso permite detectar errores, valores perdidos, valores atípicos, inconsistencias y diferencias de escala antes de tomar decisiones o construir modelos.

$$\text{Fuentes de datos} \;\xrightarrow{\text{Procesamiento de datos}}\; \text{Datos limpios y estructurados} \;\rightarrow\; \text{Análisis y conocimiento}$$

### Historia breve del Procesamiento de Datos

| Época | Hito |
|---|---|
| **1960s–70s** | Bases de datos y procesamiento por lotes para organizar grandes volúmenes de información |
| **1980s** | Sistemas de información y herramientas de consulta para apoyar decisiones |
| **1990s** | Almacenes de datos, integración de fuentes y procesos ETL |
| **2000s** | Expansión de datos web, sensores y formatos estructurados como XML y JSON |
| **2010s** | Big Data, procesamiento distribuido y análisis reproducible con Python |
| **2020+** | Automatización, datos en tiempo real, inteligencia artificial y gobierno de datos |

## 2. Importancia y Aplicaciones

El Procesamiento de Datos es una etapa **transversal** en proyectos científicos, empresariales y tecnológicos. Permite integrar fuentes distintas, evaluar la calidad de la información y producir conjuntos de datos confiables para el análisis y la toma de decisiones.

### ¿Por qué es importante?

- **Integra fuentes heterogéneas**: archivos, bases de datos, APIs, sensores e imágenes.
- **Mejora la calidad**: identifica errores, duplicados, inconsistencias y valores perdidos.
- **Facilita la comprensión**: resume distribuciones, rangos, relaciones y patrones.
- **Prepara los datos para el análisis**: transforma formatos y escalas de manera reproducible.
- **Reduce riesgos**: evita conclusiones basadas en datos incompletos o mal interpretados.
- **Aumenta la reproducibilidad**: documenta cada paso desde la extracción hasta el resultado.

### Finanzas y Banca

- Integración de transacciones provenientes de distintos sistemas.
- Detección de operaciones duplicadas, incompletas o atípicas.
- Preparación de datos para análisis de riesgo y fraude.

### Salud

- Consolidación de registros clínicos y resultados de laboratorio.
- Tratamiento de datos faltantes y estandarización de unidades.
- Organización de imágenes y señales biomédicas para su análisis.

### Retail y Marketing

- Integración de ventas, clientes, inventarios y campañas.
- Limpieza de catálogos y normalización de categorías.
- Análisis de comportamiento y demanda.

### Industria y sensores

- Recolección y sincronización de mediciones de sensores.
- Filtrado de ruido y detección de anomalías.
- Preparación de series de tiempo para monitoreo y mantenimiento.

### Geografía e imágenes

- Lectura y combinación de archivos SHP, raster e imágenes.
- Reescalamiento, análisis de componentes y extracción de información espacial.
- Elaboración de datos listos para análisis territorial.

| Aplicación | Fuentes comunes | Resultado del procesamiento |
|---|---|---|
| Riesgo financiero | Transacciones, clientes, historial | Tabla validada y consistente |
| Salud | Expedientes, laboratorios, imágenes | Datos integrados y estandarizados |
| Retail | Ventas, inventarios, campañas | Dataset limpio para análisis |
| Industria | Sensores y registros temporales | Serie sincronizada y filtrada |
| Análisis territorial | SHP, raster e imágenes | Información espacial estructurada |

> **Idea clave:** los datos rara vez llegan listos para analizar. El Procesamiento de Datos es el puente entre las fuentes originales y la información confiable.

## 3. Conceptos Clave

### 📦 Dato e información

Un **dato** es una observación, medida o registro sobre un fenómeno. La **información** surge cuando los datos se organizan, limpian y analizan dentro de un contexto.

### 📥 Fuentes y formatos de datos

Los datos pueden presentarse en distintos formatos y requerir herramientas específicas de extracción:

| Fuente o formato | Ejemplo de contenido |
|---|---|
| **Texto** | Registros, configuraciones y documentos |
| **CSV** | Tablas delimitadas por comas |
| **XLS/XLSX** | Hojas de cálculo y reportes |
| **JSON/XML** | Datos estructurados e intercambio entre sistemas |
| **SHP** | Información geoespacial vectorial |
| **Imágenes** | Fotografías, mapas, documentos y matrices de píxeles |

### 🏷️ Tipos de datos

Clasificar los datos es un paso esencial para decidir cómo explorarlos y transformarlos.

| Tipo | Descripción | Ejemplo |
|---|---|---|
| **Cuantitativo continuo** | Valores medibles dentro de un intervalo | Ingreso, temperatura |
| **Cuantitativo discreto** | Conteos expresados con números enteros | Número de visitas |
| **Categórico nominal** | Categorías sin orden | Ciudad, color |
| **Categórico ordinal** | Categorías con un orden | Nivel educativo, talla |
| **Binario** | Dos valores posibles | Sí/No |
| **Temporal** | Fechas, horas y períodos | Fecha de compra |
| **Texto** | Cadenas no estructuradas | Comentarios, reseñas |

### 📚 Dataset

Un **dataset** es una colección organizada de observaciones. Generalmente, las filas representan registros y las columnas representan variables o atributos.

| Elemento | Descripción |
|---|---|
| **Observación** | Registro individual dentro del conjunto |
| **Variable** | Atributo medido o descrito |
| **Valor** | Dato que toma una variable en una observación |
| **Metadato** | Información sobre origen, tipo, unidad y significado |

### 🧩 Problemas frecuentes de calidad

| Concepto | Definición |
|---|---|
| **Valores perdidos** | Datos ausentes, nulos o no registrados |
| **Valores atípicos** | Observaciones alejadas del comportamiento habitual |
| **Duplicados** | Registros repetidos total o parcialmente |
| **Inconsistencias** | Diferencias de formato, unidades o categorías |
| **Rango inválido** | Valores fuera de los límites esperados |
| **Sesgo** | Distribución asimétrica de los valores |
| **Desbalance** | Proporciones muy distintas entre categorías |

> **Regla de oro:** antes de transformar o analizar, hay que conocer el origen, el significado y la calidad de los datos.

## 4. Elementos y Técnicas Clave

El Procesamiento de Datos reúne técnicas que se aplican de forma ordenada según el origen, el tipo y la calidad de los datos.

### 📥 Extracción de datos

La extracción consiste en leer datos desde diferentes fuentes y convertirlos a estructuras que puedan analizarse:

| Fuente | Herramientas o técnicas |
|---|---|
| **Texto** | Lectura de archivos y separación de registros |
| **CSV** | `pandas.read_csv()` y validación de delimitadores |
| **XLS/XLSX** | `pandas.read_excel()` y selección de hojas |
| **JSON/XML** | Lectura de estructuras jerárquicas y conversión a tablas |
| **SHP** | `geopandas` y manejo de geometrías |
| **Imágenes** | Lectura de píxeles, canales y metadatos |

### 🔎 Comprensión y exploración

Antes de modificar los datos es necesario conocer su estructura:

- Revisar dimensiones, nombres de columnas y tipos de datos.
- Clasificar variables categóricas y cuantitativas.
- Examinar rangos, distribuciones y relaciones.
- Identificar valores perdidos, duplicados y atípicos.
- Documentar unidades, categorías y reglas de validación.

### 🧹 Limpieza de datos

La limpieza corrige o registra problemas que afectan la confiabilidad del análisis:

- Tratamiento de datos perdidos mediante eliminación o imputación.
- Eliminación de duplicados y corrección de formatos.
- Homologación de categorías, fechas y unidades.
- Detección y tratamiento de valores atípicos.
- Validación de rangos y reglas de negocio.

### 🔄 Transformación de datos

La transformación adapta los datos al análisis que se desea realizar:

| Técnica | Propósito |
|---|---|
| **Estandarización** | Centrar y escalar variables numéricas |
| **Normalización** | Llevar los valores a un rango comparable |
| **Codificación** | Representar categorías de forma numérica |
| **Agregación** | Resumir registros por grupos o períodos |
| **Reestructuración** | Convertir datos entre formatos ancho y largo |
| **Generación** | Crear registros o variables derivadas con reglas documentadas |

Ejemplo de estandarización:

$$z = \frac{x - \mu}{\sigma}$$

### 🖼️ Procesamiento de imágenes

Las imágenes pueden tratarse como matrices de datos:

- Cambio de tamaño y reescalamiento.
- Conversión entre espacios de color y canales.
- Filtrado, recorte y ajuste de contraste.
- Análisis de componentes para reducir o resumir información.

### 📊 Preparación para el análisis

Una vez procesados, los datos pueden organizarse para análisis descriptivo, visualización, pruebas estadísticas o modelos de aprendizaje automático. La construcción y selección de características forman parte de esta etapa posterior, pero el objetivo principal del curso es garantizar que los datos de entrada sean confiables, claros y reproducibles.

## 5. Pipeline Típico de Procesamiento de Datos

Un proyecto de datos sigue una secuencia de etapas para convertir fuentes originales en información confiable. El orden ayuda a mantener la trazabilidad, evitar errores y producir resultados reproducibles.

```mermaid
flowchart TB
    P1["1 · Fuentes de datos<br/><small>Archivos, bases de datos, APIs, sensores, imágenes</small>"]:::data
    P2["2 · Extracción<br/><small>Lectura, formatos, metadatos e integración</small>"]:::step
    P3["3 · Comprensión y exploración<br/><small>Tipos, rangos, distribuciones y relaciones</small>"]:::step
    P4["4 · Limpieza y validación<br/><small>Faltantes, duplicados, atípicos e inconsistencias</small>"]:::core
    P5["5 · Transformación<br/><small>Codificación, estandarización, normalización y agregación</small>"]:::step
    P6["6 · Construcción y selección<br/><small>Variables derivadas, reducción y preparación</small>"]:::step
    P7["7 · Análisis o modelado<br/><small>Visualización, estadística y aprendizaje automático</small>"]:::model
    P8["8 · Comunicación y mantenimiento<br/><small>Resultados, documentación y actualización</small>"]:::model

    P1 --> P2 --> P3 --> P4 --> P5 --> P6 --> P7 --> P8

    classDef data fill:#f4d35e,stroke:#b8860b,stroke-width:1px,color:#1a3a5c;
    classDef step fill:#eaf2fb,stroke:#4a9eda,stroke-width:1px,color:#1a3a5c;
    classDef core fill:#2d6a9f,stroke:#1a3a5c,stroke-width:2px,color:#ffffff;
    classDef model fill:#d8f3dc,stroke:#2d6a4f,stroke-width:1px,color:#1a3a5c;
```

### Consideraciones importantes

1. **El orden importa**: extraer y comprender antes de limpiar y transformar.
2. **Conservar los datos originales**: trabajar sobre copias o capas de procesamiento.
3. **Documentar supuestos**: registrar cambios de unidades, categorías y reglas.
4. **Evitar fugas de información**: separar correctamente los datos usados para preparar y evaluar.
5. **Validar cada etapa**: comprobar dimensiones, tipos, rangos y resultados intermedios.
6. **Hacer el proceso reproducible**: automatizar pasos y controlar versiones.
7. **Proteger la información**: respetar privacidad, permisos y seguridad de los datos.
