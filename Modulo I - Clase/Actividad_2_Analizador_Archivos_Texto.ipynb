1. Construya un analizador que recorra varios archivos `.txt`, extraiga información mediante expresiones regulares y genere un reporte consolidado. Considere los archivos disponibles en `Data`, entre ellos `texto_1.txt`, `texto_2.txt`, `texto_3.txt`, etc...

1. Define una función llamada `analizar_archivo(ruta_archivo)` que reciba la ruta de un archivo `.txt` y devuelva un diccionario con:
   - nombre del archivo;
   - número de líneas;
   - número de palabras;
   - cantidad de números encontrados;
   - cantidad de números decimales;
   - cantidad de palabras que comienzan con vocal.
   
2. Utiliza `with open(..., encoding='utf-8')` para leer cada archivo. Recorre únicamente los archivos que tengan extensión `.txt`. Ordénalos por nombre para que el análisis sea reproducible.

3. Define y aplica los siguientes patrones con el módulo `re`:
   - **Número entero o decimal:** `r'(?<![\w.])-?\d+(?:[.,]\d+)?(?![\w.])'`
   - **Número decimal:** `r'(?<![\w.])-?\d+[.,]\d+(?![\w.])'`
   - **Palabra que comienza con vocal:** `r'\b[aeiouáéíóúü][a-záéíóúüñ]*\b'`, usando `re.IGNORECASE`.
4. Convierte los resultados en un `DataFrame` y muestra una fila por archivo.


5. Conteste las siguientes Preguntas 
    1. ¿Qué archivo contiene más líneas y cuál contiene más palabras?
    2. ¿Cuántos números enteros o decimales se encontraron en total?
    3. ¿Qué archivo contiene más números decimales?
    4. ¿Qué archivo tiene la mayor proporción de palabras que contiene vocales?


```python
#Código aquí

import re

patron_numero = r'(?<![\w.])-?\d+(?:[.,]\d+)?(?![\w.])'
patron_decimal = r'(?<![\w.])-?\d+[.,]\d+(?![\w.])'
patron_vocal = r'\b[aeiouáéíóúü][a-záéíóúüñ]*\b'

def analizar_archivo(ruta_archivo):

    with open(ruta_archivo, 'r', encoding='utf-8') as archivo:
        contenido = archivo.read()

    nomb_archivo = ruta_archivo.split('\\')[-1].split('.')[0]

    cant_num = len(re.findall(patron_numero, contenido))
    num_palabras = len(re.findall(r'\b\w+\b', contenido))
    cant_num_dec = len(re.findall(patron_decimal, contenido))
    cant_vocales = len(re.findall(patron_vocal, contenido, flags = re.I))
    


    #dic_reporte_txt = {'Archivo': nomb_archivo, 'num_lineas': , 'num_palabras': , 'cantidad_num': , 'cantidad_num_dec': , 'cantidad_palabras_voc': }
    dic_reporte_txt = {'Archivo': nomb_archivo, 'num_palabras': num_palabras, 'cantidad_num': cant_num, 'cantidad_num_dec': cant_num_dec, 'cantidad_palabras_vocales' : cant_vocales}
    return dic_reporte_txt

    
```
