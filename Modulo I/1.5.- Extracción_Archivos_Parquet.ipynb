# Archivos Parquet

**Parquet** es un formato de almacenamiento **columnar** y **binario** diseñado para el análisis de datos a gran escala. A diferencia de un CSV (que guarda los datos por filas y en texto plano), Parquet organiza la información por columnas, lo que ofrece varias ventajas:

- **Compresión eficiente:** al agrupar datos del mismo tipo en cada columna, se comprime mucho mejor y ocupa menos espacio en disco.
- **Lectura selectiva:** permite leer solo las columnas necesarias sin cargar todo el archivo, acelerando las consultas.
- **Conserva los tipos de dato:** guarda el *esquema* (el tipo de cada columna), a diferencia del CSV donde todo es texto.
- **Estándar en Big Data:** es el formato preferido en herramientas como Apache Spark, Hadoop, AWS Athena y en los *data lakes*.

**¿Cuándo se usa?** Al trabajar con grandes volúmenes de datos, en flujos de trabajo analíticos y cuando se busca eficiencia en almacenamiento y velocidad de lectura.

En `pandas` se usa `df.to_parquet()` para escribir y `pd.read_parquet()` para leer. Requiere un motor instalado: `pyarrow` (recomendado) o `fastparquet`.

```python
# Configuración común para ejecutar esta sección de forma independiente
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dir_base = os.getcwd()
print(dir_base)
ruta = os.path.join(dir_base, 'Data')
print(ruta)
```

```python
# Parquet requiere un motor de lectura/escritura. Descomenta si es necesario:
# %pip install pyarrow
!pip install pyarrow
```

```python
# Creamos un DataFrame de ejemplo y lo guardamos en formato Parquet
df_parquet = pd.DataFrame({
    'producto': ['manzana', 'pera', 'plátano', 'naranja'],
    'precio': [3.5, 5.0, 2.0, 7.5],
    'cantidad': [10, 8, 20, 5]
})
#Guardando en formato parquet
```

```python
df_parquet.to_parquet(ruta+'/productos.parquet')
```

```python
# Lectura completa del archivo Parquet
pd.read_parquet(ruta+'/productos.parquet')
```

```python
# Ventaja de Parquet: leer únicamente algunas columnas
df_p_leidos = pd.read_parquet(ruta+'/productos.parquet', columns = ["precio", "cantidad"])
```

```python
# El esquema (tipos) se conserva, y el tamaño en disco suele ser menor que un CSV
df_p_leidos.dtypes
#Crear un csv


#Verificación de tamaño de csv y parquet
```

```python
df_parquet.to_csv(ruta+'/productos.parquet')
```
