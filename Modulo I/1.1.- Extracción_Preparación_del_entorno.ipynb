## El módulo `os` para gestionar rutas y archivos

El módulo estándar `os` permite interactuar con el sistema operativo: consultar el directorio de trabajo, listar archivos, construir rutas portables y obtener información de cada archivo. Usarlo hace que el cuaderno funcione igual en Windows, Linux o macOS, sin depender de rutas escritas "a mano".

`os.path` es especialmente útil para trabajar con rutas: `os.path.join` las construye de forma segura y `os.path.exists` verifica si un archivo existe antes de intentar abrirlo.

#### Obtener el directorio actual

Para obtener el directorio de trabajo actual, usamos el función `getcwd()` (_get current working directory_) del módulo `os`.

Esta función te devuelve el directorio actual en forma de cadena.

````python
# Directorio de trabajo actual (desde donde se ejecuta el cuaderno)
print('Directorio actual:', os.getcwd())
````

Es importante ver que la salida dependerá del sistema operativo que estés usando. Por ejemplo, en Windows podrías obtener algo así: `C:\\usuario\\ejercicios_python`.

#### Listar directorios y archivos

La función `listdir()` toma un directorio (_path_ o camino) y devuelve una lista con todos los archivos y subdirectorios de un directorio. Si no se le pasa ningún parámetro, devuelve los del directorio de trabajo actual.


````python
# Listar todos los archivos disponibles en la carpeta Data
os.listdir(ruta)
````

La función `rename()` también es útil para mover un archivo o directorio, cambiando el camino (_path_) de acceso al archivo. Probemos hacer lo siguiente:

````python
os.rename(os.path.join('test','carpeta_vieja'), 'carpeta_vieja') # cambio el path
print(os.listdir('test'))
````

La carpeta 'carpeta' ahora se encuentra en 'Modulo I', y no dentro de 'Modulo I/test'.

**Ojo**: `rename()` funciona cuando el archivo (o directorio) no se cambia de disco (o más específicamente de una partición). Si se necesita mover un archivo del disco a un pendrive, por ejemplo, lo correcto es copiar el archivo al pendrive y luego borrarlo del disco. `rename()` no hace esto: no copia y borra, simplemente cambia el nombre. Para renombrar en caso que se pueda o copiar y borrar si lo primero no es posible, podemos usar la función `move()` del módulo `shutil`. Este módulo es de más alto nivel y usa las primitivas de bajo nivel del módulo `os`. Al usar `os` tenemos un control más estricto de las operaciones. Las funciones de `shutil` pueden resultar más cómodas, pero en el camino pueden invocar a diversas funciones de bajo nivel del módulo `os`.

#### Recorriendo directorios con `os.walk()`

La función `walk()` del módulo `os` genera una lista con los nombres de todos los archivos del árbol de subdirectorios de un directorio dado. Es decir, lista los archivos de un directorio dado y luego entra en cada subdirectorio y hace lo mismo, recursivamente (_top-down_).

La función `walk()` recibe como único parámetro obligatorio el directorio donde comenzar a mirar (la raíz del árbol).

##### Ejemplo

En este ejemplo se ve cómo se usa `os.walk()`.

```python
import os
for root, dirs, files in os.walk("."):
   for name in files:
      print(os.path.join(root, name))
   for name in dirs:
      print(os.path.join(root, name))
```

`os.walk` recorre un directorio junto con todos sus subdirectorios. Es útil para localizar archivos en estructuras de carpetas anidadas.

````python
directorio = os.path.join('c:\\', 'Users','uie70742','OneDrive - ITESO','ITESO', 'Ingeniería y Ciencia de Datos','Laboratorio_de_Procesamiento_de_Datos_02026','Modulo I')
os.chdir(directorio)
print(os.getcwd())
````

````python
# Recorrer la carpeta Data y sus subcarpetas
for carpeta_actual, subcarpetas, archivos in os.walk(ruta):
    print('Carpeta:', carpeta_actual)
    print('  Subcarpetas:', subcarpetas)
    print('  Archivos:', archivos)
````
