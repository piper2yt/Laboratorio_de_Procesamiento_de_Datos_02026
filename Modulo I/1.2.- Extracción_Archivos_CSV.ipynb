# Archivos CSV

Los archivos CSV representan datos tabulares: cada fila suele corresponder a una observación y cada columna a una variable. Son frecuentes en reportes, exportaciones de sistemas y conjuntos de datos pequeños o medianos.

`pandas` permite cargarlos en un `DataFrame`, una estructura que facilita filtrar, resumir y transformar datos. La lectura correcta depende de detalles como el separador, la codificación, los encabezados y los valores faltantes.

**Checklist de inspección:** después de cargar un archivo revisa `shape`, `columns`, `dtypes`, valores nulos y algunas filas. En el ejemplo se elimina `Unnamed: 0`, una columna índice que suele aparecer cuando un DataFrame se exportó sin `index=False`.

### Leer un CSV con `with open` y el módulo `csv`

Antes de usar `pandas` conviene entender qué ocurre "por debajo". Un CSV es simplemente **texto plano**: cada línea es una fila y las columnas se separan con un delimitador (normalmente una coma). Con `open` y el módulo estándar `csv` se puede leer sin dependencias externas.

- `csv.reader` devuelve cada fila como una **lista** de cadenas.
- `csv.DictReader` usa la primera fila como encabezado y devuelve cada fila como un **diccionario**.

````python
import csv

# Lectura manual del CSV con csv.reader (cada fila es una lista)
with open(ruta + '/df_tabla2.csv', newline='', encoding='utf-8') as f:
    lector = csv.reader(f)
    for fila in lector:
        print(fila)
````

````python
# csv.DictReader usa la primera fila como encabezado y devuelve diccionarios
with open(ruta + '/df_tabla2.csv', newline='', encoding='utf-8') as f:
    lector = csv.DictReader(f)
    filas = list(lector)
````

#### Comparación: `with open` + `csv` vs `pd.read_csv`

| Aspecto | `with open` + `csv` | `pd.read_csv` |
|---|---|---|
| Resultado | listas o diccionarios de Python | `DataFrame` |
| Tipos de dato | todo se lee como texto (`str`) | infiere números, fechas, etc. |
| Valores faltantes | se manejan a mano | se convierten a `NaN` |
| Filtrado y análisis | requiere código adicional | métodos integrados |
| Dependencias | solo librería estándar | requiere `pandas` |
| Archivos grandes | control fila por fila | cómodo, pero carga todo en memoria |

En resumen, `with open` + `csv` ofrece control total sin librerías externas, mientras que `pd.read_csv` es más práctico para el análisis de datos porque entrega directamente un `DataFrame` con los tipos ya inferidos.

````python
# Con csv todo es texto; pd.read_csv infiere los tipos automáticamente
df_comp = pd.read_csv(ruta + '/df_tabla2.csv')
print('Tipos inferidos por pandas:')
print(df_comp.dtypes)
````
